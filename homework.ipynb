```python
import math
print("Введите длину первой стороны: ")
s1 = input()
print("Введите длину второй стороны: ")
s2 = input()
print("Введите измерения угла между сторонами: ")
u = input()
a = math.sqrt((int(s1)**2.0)+(int(s2)**2.0)-(2.0*int(s1)*int(s2)*(math.cos(int(u)))))
print("Ответ: ", a)
```
