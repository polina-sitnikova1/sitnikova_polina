```python
print("Введите температуру (например, '40C' или '104F'): ")
t = input()

# f = (c*1.8) + 32
# c = (f - 32)/1.8

if t == "":
    print("Введите температуру, как указано в примере")

elif "C" in t:
    inpc = list(t)
    del inpc[-1]
    resc = ""
    for i in inpc:  #???
        resc += str(i)
        
    resc = float(resc)
    f = (resc*1.8) + 32
    f = str(f)
    print(f + "F")
    
elif "F" in t:
    inpf = list(t)
    del inpf[-1]
    resf = ""
    for i in inpf:  #???
        resf += str(i)
        
    resf = float(resf)
    c = (resf - 32)/1.8
    c = str(c)
    print(c + "C")

else:
    print("Введите температуру, как указано в примере")


```
