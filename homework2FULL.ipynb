```python
print("Введите число от 0 до 99: ")
n = input()

if n == "": # если пустота, то введите число
    print("Введите число")
elif not n.isdigit():  # проверяет, состоят ли все символы в строке исключительно из цифровых символов. вызов - str.isdigit()
    print("Введите число")
else:
    n = int(n)
    res = ""

    if len(str(n)) == 1: # если однозначное число

        if int(n) == 0: res = "ноль" 
        elif int(n) == 1: res = "один"
        elif int(n) == 2: res = "два"
        elif int(n) == 3: res = "три"
        elif int(n) == 4: res = "четыре"
        elif int(n) == 5: res = "пять"
        elif int(n) == 6: res = "шесть"
        elif int(n) == 7: res = "семь"
        elif int(n) == 8: res = "восемь"
        elif int(n) == 9: res = "девять"

    elif len(str(n)) == 2: # если двузначное число

        if int(n) == 10: res = "десять"
        elif int(n) == 11: res = "одиннадцать"
        elif int(n) == 12: res = "двенадцать"
        elif int(n) == 13: res = "тринадцать"
        elif int(n) == 14: res = "четырнадцать"
        elif int(n) == 15: res = "пятнадцать"
        elif int(n) == 16: res = "шестнадцать"
        elif int(n) == 17: res = "семнадцать"
        elif int(n) == 18: res = "восемнадцать"
        elif int(n) == 19: res = "девятнадцать"
        
        if int(n) // 10 == 2: res = "двадцать" 
        elif int(n) // 10 == 3: res = "тридцать"
        elif int(n) // 10 == 4: res = "сорок"
        elif int(n) // 10 == 5: res = "пятьдесят"
        elif int(n) // 10 == 6: res = "шестьдесят"
        elif int(n) // 10 == 7: res = "семьдесят"
        elif int(n) // 10 == 8: res = "восемьдесят"
        elif int(n) // 10 == 9: res = "девяносто"

        if int(n) > 19: # конкантинируем строки, создавая двузначные числа, состоящие из нескольких слов 
            if int(n) % 10 == 1: res = res + " один"
            elif int(n) % 10 == 2: res = res + " два"
            elif int(n) % 10 == 3: res = res + " три"
            elif int(n) % 10 == 4: res = res + " четыре"
            elif int(n) % 10 == 5: res = res + " пять"
            elif int(n) % 10 == 6: res = res + " шесть"
            elif int(n) % 10 == 7: res = res + " семь"
            elif int(n) % 10 == 8: res = res + " восемь"
            elif int(n) % 10 == 9: res = res + " девять"

    print(res) # печатаем
```
